**Project Stage:** Review 2 Prototype Evaluation  
**Notice:** This notebook operates exclusively on **synthetic patient data** (`adult_001` through `adult_010`). Visualisations and statistics are non-diagnostic and provided solely for technical evaluation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys

# Ensure project root is in python path
sys.path.append('..')
from src.data_generator import save_raw_dataset
from src.data_cleaning import run_cleaning_pipeline

# Ensure data exists and is cleaned
cleaned_df = run_cleaning_pipeline()
cleaned_df.head()
```

## 1. Dataset Shape & Column Schema

```python
print(f"Dataset Shape: {cleaned_df.shape[0]} rows x {cleaned_df.shape[1]} columns")
print("\nColumn Names & Data Types:")
print(cleaned_df.dtypes)
```

## 2. Missing Values & Summary Statistics

```python
print("Missing Value Count per Column:")
print(cleaned_df.isna().sum())

print("\nDescriptive Statistics for Vital Signs:")
cleaned_df[['heart_rate', 'systolic_bp', 'diastolic_bp', 'oxygen_saturation', 'temperature']].describe()
```

## 3. Visualisations: Vitals Over Time
Comparing Heart Rate, Systolic BP, and Oxygen Saturation over 30 days across synthetic older adult profiles.

```python
cleaned_df['timestamp'] = pd.to_datetime(cleaned_df['timestamp'])

fig, axes = plt.subplots(3, 1, figsize=(12, 12), sharex=True)

# Plot Heart Rate
for pid in ['adult_001', 'adult_005', 'adult_007', 'adult_009']:
    pdata = cleaned_df[cleaned_df['person_id'] == pid]
    axes[0].plot(pdata['timestamp'], pdata['heart_rate'], label=pid, marker='o', markersize=3)
axes[0].set_title('1. Heart Rate over Time (bpm)')
axes[0].set_ylabel('Heart Rate')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot Systolic BP
for pid in ['adult_001', 'adult_005', 'adult_007', 'adult_009']:
    pdata = cleaned_df[cleaned_df['person_id'] == pid]
    axes[1].plot(pdata['timestamp'], pdata['systolic_bp'], label=pid, marker='s', markersize=3)
axes[1].set_title('2. Systolic BP over Time (mmHg)')
axes[1].set_ylabel('Systolic BP')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Plot Oxygen Saturation
for pid in ['adult_001', 'adult_005', 'adult_007', 'adult_009']:
    pdata = cleaned_df[cleaned_df['person_id'] == pid]
    axes[2].plot(pdata['timestamp'], pdata['oxygen_saturation'], label=pid, marker='^', markersize=3)
axes[2].set_title('3. Oxygen Saturation over Time (%)')
axes[2].set_ylabel('SpO2 (%)')
axes[2].set_xlabel('Timestamp')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
